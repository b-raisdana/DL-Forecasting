# NOW Input Visual Review

Offline visual review notebook for already-fetched cached data. It does not call broker/exchange fetchers.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
while ROOT.name != "DL-Forecasting" and ROOT.parent != ROOT:
    ROOT = ROOT.parent
APP = ROOT / "app"
if str(APP) not in sys.path:
    sys.path.insert(0, str(APP))

from FigurePlotter.now_review_notebook import (
    SYMBOLS,
    build_review_samples,
    cache_summary,
    plot_now_sample,
    plot_random_period,
    random_cached_period,
)

SYMBOLS
```

## Local Cache Availability

```python
cache_summary(SYMBOLS)
```

## NOW Details

Each figure shows the selected normalized OHLCV input candles and faded future candles used by label generation, followed by the same sample in original OHLCV values. SL and TP1-TP4 ladders are drawn when the label action head is non-zero.

```python
for symbol in SYMBOLS:
    try:
        samples = build_review_samples(symbol, sample_count=5)
    except Exception as exc:
        print(f"{symbol}: skipped ({exc})")
        continue
    for idx, sample in enumerate(samples, start=1):
        fig = plot_now_sample(sample, title=f"{symbol} NOW sample {idx}")
        fig.show()
```

## Random Two-Week Ranges

Two cached two-week periods are selected per symbol from local files only.

```python
for symbol in SYMBOLS:
    for chart_no, seed in enumerate([11, 29], start=1):
        try:
            start, end = random_cached_period(symbol, seed=seed)
            fig = plot_random_period(symbol, start, end, chart_no)
        except Exception as exc:
            print(f"{symbol} period chart {chart_no}: skipped ({exc})")
            continue
        fig.show()
```
