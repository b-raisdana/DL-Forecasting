## Local Cache Availability

```python
cache_summary(SYMBOLS)
```

Each figure shows one Tier-1_000 model-input sample built by `datafeeder_input3_outcome1.build_dataset()`: one row per branch timeframe (5min/15min/1h/4h/1D/1W), plotting that branch's own `CANDLE_FEATURE_COLUMNS` (ATR-relative OHLC + volume) across its window, with `x=0` marking NOW. The realized `mfe`/`rer`/`action` labels (outcome_set=1) are summarized in each figure's title.

```python
for symbol in SYMBOLS:
    try:
        samples = build_review_samples(symbol, sample_count=5)
    except Exception as exc:
        print(f"{symbol}: skipped ({exc})")
        continue
    for idx, sample in enumerate(samples, start=1):
        fig = plot_now_sample(sample, title=f"{symbol} NOW sample {idx}")
        fig.show()
```

## Random Two-Week Ranges

Two cached two-week periods are selected per symbol from local files only.

```python
for symbol in SYMBOLS:
    for chart_no, seed in enumerate([11, 29], start=1):
        try:
            start, end = random_cached_period(symbol, seed=seed)
            fig = plot_random_period(symbol, start, end, chart_no)
        except Exception as exc:
            print(f"{symbol} period chart {chart_no}: skipped ({exc})")
            continue
        fig.show()
```
