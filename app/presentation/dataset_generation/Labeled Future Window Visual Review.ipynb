# Labeled Future Window — Dataset Generation Visual Review

Offline visual review of Tier-1_000 `input3_outcome1` samples. Shows the **actual 5min candlesticks** around each NOW anchor, shades the **240-minute future window**, and overlays the computed **action / mfe / rer labels** on it.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
while ROOT.name != "DL-Forecasting" and ROOT.parent != ROOT:
    ROOT = ROOT.parent
APP = ROOT / "app"
if str(APP) not in sys.path:
    sys.path.insert(0, str(APP))

from presentation.dataset_generation.labeled_future_window_review import (
    SYMBOLS,
    build_review_samples,
    plot_labeled_branch_sample,
    plot_labeled_future_window,
)
```

## Candlestick review — future window with labels

Each figure shows 5min candles around a NOW anchor. Candles after NOW belong to the label horizon. The selected extremum and its MFE / RER are annotated.

```python
for symbol in SYMBOLS:
    try:
        samples = build_review_samples(symbol, sample_count=5)
    except Exception as exc:
        print(f"{symbol}: skipped ({exc})")
        continue
    for idx, sample in enumerate(samples, start=1):
        fig = plot_labeled_future_window(sample, title=f"{symbol} future window sample {idx}")
        fig.show()
```

## Branch feature windows (reference)

One row per branch timeframe, x=0 at NOW. Labels are summarized in each title.

```python
for symbol in SYMBOLS:
    try:
        samples = build_review_samples(symbol, sample_count=5)
    except Exception as exc:
        print(f"{symbol}: skipped ({exc})")
        continue
    for idx, sample in enumerate(samples, start=1):
        fig = plot_labeled_branch_sample(sample, title=f"{symbol} branch sample {idx}")
        fig.show()
```
